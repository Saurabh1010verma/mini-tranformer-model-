# Mini GPT Experiments Notebook

This notebook demonstrates how to interactively inspect, train, and generate text using **Mini GPT from Scratch** built with PyTorch.

```python
import torch
from mini_gpt.config import ModelConfig, Config
from mini_gpt.tokenizer import Tokenizer
from mini_gpt.model.gpt import MiniGPT
from mini_gpt.generation.generate import generate_text

print(f"PyTorch Version: {torch.__version__}")
```

## 1. Tokenizer Demo

```python
sample_text = "To be, or not to be, that is the question."
tokenizer = Tokenizer(mode="bpe")
tokenizer.fit(sample_text, target_vocab_size=100)

encoded = tokenizer.encode(sample_text)
decoded = tokenizer.decode(encoded)

print(f"Original: {sample_text}")
print(f"Token IDs: {encoded}")
print(f"Decoded:  {decoded}")
```

## 2. Inspect Model Architecture

```python
config = ModelConfig(
    vocab_size=1000,
    context_length=256,
    embedding_dim=256,
    num_heads=8,
    num_layers=6,
    dropout=0.1
)
model = MiniGPT(config)
print(model)
print(f"Total Non-Embedding Parameters: {model.get_num_params():,}")
```

## 3. Verify Forward Pass & Causal Masking

```python
x = torch.randint(0, 1000, (2, 32))  # Batch size 2, Sequence length 32
y = torch.randint(0, 1000, (2, 32))

logits, loss = model(x, targets=y)
print(f"Logits Shape: {logits.shape}")
print(f"Calculated Cross-Entropy Loss: {loss.item():.4f}")
```
